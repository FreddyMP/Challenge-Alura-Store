### Importación de datos



```python
import pandas as pd
from matplotlib import pyplot as plt

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
total = tienda['Precio'].sum()
total2 = tienda2['Precio'].sum()
total3 = tienda3['Precio'].sum()
total4 = tienda4['Precio'].sum()

print(f"Tienda 1 un total de: {total:,.2f} \nTienda 2 un total de: {total2:,.2f} \nTienda 3 un total de: {total3:,.2f}\nTienda 4 un total de: {total4:,.2f}")

Tiendas = ["Tienda1", "Tienda2", "Tienda3","Tienda4"]
Ventas = [total,total2, total3,total4  ]

mayorVentaIndice = Ventas.index(max(Ventas))
menorVentaIndice = Ventas.index(min(Ventas))


print(f"\nLa tienda que mas vendio fue la {Tiendas[mayorVentaIndice]} \nLa que menos vendio fue la {Tiendas[menorVentaIndice]} ")

plt.bar(x=Tiendas, height=Ventas)
plt.show()
```

# 2. Ventas por categoría

```python
tiendasCategorias = [tienda['Categoría del Producto'].sum(),tienda2['Categoría del Producto'].sum(),tienda3['Categoría del Producto'].sum(),tienda4['Categoría del Producto'].sum()]
tiendasList = [tienda,tienda2,tienda3,tienda4]
tiendasListName = ["tienda","tienda 2","tienda3","tienda4"]
x = 0
while(x < len(tiendasCategorias)):
  categoriasAll = tiendasCategorias[x]
  categoriasClear = list(set(categoriasAll))

  ventas_por_categoria = tiendasList[x].groupby("Categoría del Producto")["Precio"].sum()

  print(ventas_por_categoria)

  ventas_por_categoria.plot(kind="bar", color="skyblue", edgecolor="black")

  plt.title(f"Ventas por Categoría {tiendasListName[x]}")
  plt.xlabel("Categoría del Producto")
  plt.ylabel("Ventas Totales (COP)")
  plt.xticks(rotation=60)
  plt.tight_layout()
  plt.show()

  x = x + 1


```

# 3. Calificación promedio de la tienda


```python
tiendasCalificacionTotal = [tienda['Calificación'].sum(),tienda2['Calificación'].sum(),tienda3['Calificación'].sum(),tienda4['Calificación'].sum()]
tiendasCalificacion = [tienda['Calificación'],tienda2['Calificación'],tienda3['Calificación'],tienda4['Calificación']]
tiendasList = ["Tienda 1","Tienda 2","Tienda 3","tienda 4"]
tiendasListPromedio = []

x = 0
while(x < len(tiendasCalificacionTotal)):

  sumaCalificacion = tiendasCalificacionTotal[x]
  cantidadCalificaciones = len(tiendasCalificacion[x])
  promedio = sumaCalificacion/cantidadCalificaciones

  tiendasListPromedio.append(promedio)

  print(f"Promedio de la {tiendasList[x]} es: {promedio}")

  x = x + 1

plt.bar(x=tiendasList, height=tiendasListPromedio)
plt.show()

```

# 4. Productos más y menos vendidos

```python
tiendasCategorias = [tienda['Producto'].sum(),tienda2['Producto'].sum(),tienda3['Producto'].sum(),tienda4['Producto'].sum()]
tiendasList = [tienda,tienda2,tienda3,tienda4]
tiendasListName = ["tienda","tienda 2","tienda3","tienda4"]
x = 0
while(x < len(tiendasCategorias)):
  categoriasAll = tiendasCategorias[x]
  categoriasClear = list(set(categoriasAll))

  ventas_por_categoria = tiendasList[x].groupby("Producto")["Precio"].sum()

  print(ventas_por_categoria)

  ventas_por_categoria.plot(kind="bar", color="skyblue", edgecolor="black")

  plt.title(f"Ventas por Producto {tiendasListName[x]}")
  plt.xlabel("Producto")
  plt.ylabel("Ventas Totales (COP)")
  plt.show()

  x = x + 1
```

# 5. Envío promedio por tienda

```python
tiendasEnvioTotal = [tienda['Costo de envío'].sum(),tienda2['Costo de envío'].sum(),tienda3['Costo de envío'].sum(),tienda4['Costo de envío'].sum()]
tiendasEnvio = [tienda['Costo de envío'],tienda2['Costo de envío'],tienda3['Costo de envío'],tienda4['Costo de envío']]
tiendasList = ["Tienda 1","Tienda 2","Tienda 3","tienda 4"]
tiendasListPromedio = []

x = 0
while(x < len(tiendasEnvioTotal)):

  sumaEnvio = tiendasEnvioTotal[x]
  cantidadEnvio = len(tiendasEnvio[x])
  promedio = sumaEnvio/cantidadEnvio

  tiendasListPromedio.append(promedio)

  print(f"Promedio de la {tiendasList[x]} es: {promedio}")

  x = x + 1

plt.bar(x=tiendasList, height=tiendasListPromedio)
plt.show()
```
